# Second policy: train ManiSkill's Diffusion Policy baseline
Trains the official ManiSkill 3 Diffusion Policy (RGB, 2 frames of the 128x128 sensor camera, U-Net denoiser,
100 DDPM steps) on 100 motion-planning demos per task, with the settings of ManiSkill's `baselines.sh`
(30k iterations for PickCube / PushCube). Checkpoints are saved to Drive next to the RDT results. StackCube is left
out by default: RGB Diffusion Policy is reported to fail on it.

A training run does not resume after a disconnect (it restarts that task); finished tasks are skipped.

```python
#@title Train the second policy: ManiSkill's Diffusion Policy baseline (RGB). Any NVIDIA GPU works (L4 / T4 too).
OUT_DRIVE = '/content/drive/MyDrive/rdt-xai/pilot2'  #@param {type:"string"}
DP_TASKS = ['PickCube-v1', 'PushCube-v1']  #@param
import os, sys, subprocess, importlib
from google.colab import drive
drive.mount('/content/drive')
REPO = '/content/rdt-igtesting'
if not os.path.exists(REPO):
    subprocess.run(f'git clone -q -b cvpr-xai https://github.com/rdgbrandon/rdt-igtesting {REPO}', shell=True, check=True)
subprocess.run(f'cd {REPO} && git pull -q origin cvpr-xai', shell=True, check=True)
sys.path.insert(0, REPO)
import rdt_xai.colab as colab
importlib.reload(colab)
colab.prepare()                              # installs ManiSkill (current) etc.; no RDT model download needed
for task in DP_TASKS:
    print(f'\n===== training Diffusion Policy on {task} =====', flush=True)
    colab.train_dp(task, OUT_DRIVE)          # checkpoints -> <OUT_DRIVE>/dp/<task>/ ; skipped if already trained
```
